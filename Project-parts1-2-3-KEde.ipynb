````python
#Your code goes here below
# PART-ONE: Marking Activities and Habits as Complete
# Import the library feature used below.
from datetime import datetime

# Function marking activities completed starts HERE below
# Define the function and its inputs.
def markActivityComplete():
    # Collect all incomplete activities
    # Create an empty list to collect results.
    incomplete_activities = []

    # Loop through the items in this collection.
    for weekday in weekDays:
        # Loop through the items in this collection.
        for activity_name, completed in activities[weekday].items():
            # Check whether this condition is true.
            if not completed:
                # Add this item to the results list.
                incomplete_activities.append((weekday, activity_name))

    # Check if there are incomplete activities
    # Check whether this condition is true.
    if not incomplete_activities:
        # Display a message or result.
        print("All activities completed!")
        # Stop the function and return control to its caller.
        return

    # Display incomplete activities
    # Display a message or result.
    print("\nIncomplete Activities:")
    # Loop through the items in this collection.
    for weekday, activity_name in incomplete_activities:
        # Display a message or result.
        print(f"- {activity_name} ({weekday})")

    # Ask which activity was completed
    # Read and clean the value entered by the user.
    activity_name = input("Please kindly enter the completed activity name: ").strip().title()

    # Find the activity and update its status
    # Assign the value or calculated result to a variable.
    found = False

    # Loop through the items in this collection.
    for weekday in weekDays:
        # Check whether this condition is true.
        if activity_name in activities[weekday] and not activities[weekday][activity_name]:
            # Store or update this task or habit record.
            activities[weekday][activity_name] = True
            # Display a message or result.
            print(f"Activity '{activity_name}' completed for {weekday}")
            # Assign the value or calculated result to a variable.
            found = True

            # Bonus: Check if the activity is from an earlier weekday
            # Assign the value or calculated result to a variable.
            today = datetime.now().strftime("%A")

            # Check whether this condition is true.
            if today in weekDays:
                # Check whether this condition is true.
                if weekDays.index(weekday) < weekDays.index(today):
                    # Display a message or result.
                    print("Better late than never!")

            # Exit the current loop.
            break
        
    # Check whether this condition is true.
    if not found:
        # Display a message or result.
        print("Activity not found among incomplete activities.")

# Function marking activities completed stops HERE above
        
# Function marking habits completed start HERE below
# Define the function and its inputs.
def markHabitComplete():
    # Ask for the weekday
    # Read and clean the value entered by the user.
    weekday = input("Please kindly enter the weekday: ").strip().title()

    # Validate the weekday
    # Check whether this condition is true.
    if weekday not in weekDays:
        # Display a message or result.
        print("Invalid weekday.")
        # Stop the function and return control to its caller.
        return

    # Ask for the habit name
    # Read and clean the value entered by the user.
    habitName = input("Please kindly enter the habit name: ").strip().title()

    # Check if the habit exists
    # Check whether this condition is true.
    if habitName not in habits:
        # Display a message or result.
        print("Habit not found.")
        # Stop the function and return control to its caller.
        return

    # Update the habit status
    # Store or update this task or habit record.
    habits[habitName][weekday] = True

    # Display a message or result.
    print(f"Habit '{habitName}' completed for {weekday}!")
````

````python
#Your code goes here

# PART-TWO: REMOVING ACTIVITIES AND HABITS

# Define a function to remove an activity from a selected weekday.
def removeActivity():

    # Ask the user to enter the weekday.
    weekday = input("Please enter the weekday: ").strip().title()

    # Check whether the weekday is valid.
    if weekday not in weekDays:
        print("Invalid weekday! Please enter a valid weekday.")
        return

    # Check whether there are activities for the selected weekday.
    if not activities[weekday]:
        print("There are no activities for this weekday.")
        return

    # Display the activities available for that weekday.
    print("\nActivities for", weekday)
    for activityName in activities[weekday]:
        print("-", activityName)

    # Ask the user which activity should be removed.
    activityName = input("Enter the activity to remove: ").strip().title()

    # Check whether the activity exists in the dictionary.
    if activityName in activities[weekday]:

        # Delete the selected activity.
        del activities[weekday][activityName]

        # Display a message confirming the removal.
        print(activityName, "has been removed from", weekday)
    else:
        # Display an error message if the activity is not found.
        print("Activity not found for the selected weekday.")


# Define a function to remove a habit from the tracker.
def removeHabit():

    # Check whether the habits dictionary is empty.
    if not habits:
        print("There are no habits to remove.")
        return

    # Display all habits currently being tracked.
    print("\nAvailable habits:")
    for habitName in habits:
        print("-", habitName)

    # Ask the user to enter the habit to remove.
    habitName = input("Please, enter the habit to remove: ").strip().title()

    # Check whether the entered habit exists.
    if habitName in habits:

        # Delete the habit from the dictionary.
        del habits[habitName]

        # Display a message confirming the removal.
        print(habitName, "Habit has been removed successfully.")
    else:
        # Display a message if the habit does not exist.
        print("Habit not found.")
````

````python
#Your code goes here

# PART-TWO: GENERATING WEEKLY AND DAILY REPORTS

# Define a function to display the weekly progress report.
def reportWeek():

    # Display the heading for the weekly report.
    print("\n===== WEEKLY PROGRESS REPORT =====")

    # Display the weekly habit progress.
    print("\nHABIT PROGRESS:")

    # Check whether any habits are being tracked.
    if not habits:
        print("No habits are currently being tracked.")
    else:
        # Go through each habit and its daily completion records.
        for habitName, dailyStatus in habits.items():

            # Count how many days the habit was completed.
            completedDays = sum(1 for status in dailyStatus.values() if status)

            # Display the number of completed days out of seven.
            print(f"{habitName}: {completedDays}/7 days complete")

    # Create lists for completed and incomplete activities.
    completedActivities = []
    incompleteActivities = []

    # Go through each weekday in the tracker.
    for weekday in weekDays:

        # Go through all activities recorded for the weekday.
        for activityName, completed in activities[weekday].items():

            # Include the weekday beside the activity name.
            activityRecord = f"{activityName} ({weekday})"

            # Separate completed activities from incomplete ones.
            if completed:
                completedActivities.append(activityRecord)
            else:
                incompleteActivities.append(activityRecord)

    # Display the completed activities.
    print("\nCOMPLETED ACTIVITIES:")

    # Check whether the completed activities list contains records.
    if completedActivities:
        for activity in completedActivities:
            print("ACCEPT", activity)
    else:
        print("No completed activities recorded.")

    # Display the incomplete activities.
    print("\nINCOMPLETE ACTIVITIES:")

    # Check whether the incomplete activities list contains records.
    if incompleteActivities:
        for activity in incompleteActivities:
            print("REJECT", activity)
    else:
        print("No incomplete activities recorded.")

    # Check whether the tracker contains any activities at all.
    if not completedActivities and not incompleteActivities:
        print("\nNo activities have been recorded for the week.")


# Define a function to display the progress for one weekday.
def reportDay():

    # Ask the user which weekday they want to view.
    weekday = input("Enter the weekday for the report: ").strip().title()

    # Check whether the entered weekday is valid.
    if weekday not in weekDays:
        print("Invalid weekday! Please enter a valid weekday.")
        return

    # Display the heading for the selected day.
    print(f"\n===== DAILY REPORT FOR {weekday.upper()} =====")

    # Display the activities section.
    print("\nACTIVITIES:")

    # Check whether activities exist for the selected weekday.
    if not activities[weekday]:
        print("No activities recorded for this weekday.")
    else:
        # Go through the activities and their completion records.
        for activityName, completed in activities[weekday].items():

            # Choose the appropriate symbol based on completion.
            status = "ACCEPT" if completed else "REJECT"

            # Display the activity and its completion status.
            print(status, activityName)

    # Display the habits section.
    print("\nHABITS:")

    # Check whether any habits are being tracked.
    if not habits:
        print("No habits are currently being tracked.")
    else:
        # Go through each habit and check the selected weekday.
        for habitName, dailyStatus in habits.items():

            # Get the completion status for the selected weekday.
            completed = dailyStatus.get(weekday, False)

            # Choose the appropriate completion symbol.
            status = "ACCEPT" if completed else "REJECT"

            # Display the habit and its completion status.
            print(status, habitName)
````

````python
#Your code goes here

# PART THREE: MAIN MENU SYSTEM

# This function displays the menu and handles the user's choices.
def main():

    # Keep showing the menu until the user chooses to exit.
    while True:

        # Display the available options.
        print("\n===== WEEKLY TASK AND HABIT TRACKER =====")
        print("1. Add a Task or Habit")
        print("2. Mark a Task or Habit as Complete")
        print("3. Remove a Task or Habit")
        print("4. View Reports")
        print("5. Exit")

        # Ask the user to select an option.
        choice = input("Enter your choice (1-5): ").strip()

        # Option 1: Add an activity or habit.
        if choice == "1":
            print("\na. Add Activity")
            print("b. Add Habit")
            selection = input("Choose a or b: ").strip().lower()

            if selection == "a":
                addActivity()
            elif selection == "b":
                addHabit()
            else:
                print("Invalid choice. Please select a or b.")

        # Option 2: Mark an activity or habit as complete.
        elif choice == "2":
            print("\na. Complete Activity")
            print("b. Complete Habit")
            selection = input("Choose a or b: ").strip().lower()

            if selection == "a":
                markActivityComplete()
            elif selection == "b":
                markHabitComplete()
            else:
                print("Invalid choice. Please select a or b.")

        # Option 3: Remove an activity or habit.
        elif choice == "3":
            print("\na. Remove Activity")
            print("b. Remove Habit")
            selection = input("Choose a or b: ").strip().lower()

            if selection == "a":
                removeActivity()
            elif selection == "b":
                removeHabit()
            else:
                print("Invalid choice. Please select a or b.")

        # Option 4: Display weekly or daily progress.
        elif choice == "4":
            print("\na. Weekly Report")
            print("b. Daily Report")
            selection = input("Choose a or b: ").strip().lower()

            if selection == "a":
                reportWeek()
            elif selection == "b":
                reportDay()
            else:
                print("Invalid choice. Please select a or b.")

        # Option 5: Exit the program.
        elif choice == "5":
            print("Thank you for using the Weekly Task and Habit Tracker.")
            break

        # Handle an invalid main menu choice.
        else:
            print("Invalid option. Please enter a number from 1 to 5.")


# Start the main menu program.
main()
````
